## Feature engineering

### Subtask:
Calculate moving averages and technical indicators based on the historical price data.


**Reasoning**:
Calculate the moving averages and add the volume to the DataFrame as new columns.



```python
# Calculate moving averages
data['MA_5'] = data['Close'].rolling(window=5).mean()
data['MA_10'] = data['Close'].rolling(window=10).mean()
data['MA_50'] = data['Close'].rolling(window=50).mean()

# Display the first few rows with the new features
display(data.head())
```

## Prepare data for modeling

### Subtask:
Create the target variable (tomorrow's stock price) and select the features.


**Reasoning**:
Create the target variable, remove the last row with the missing target, and select the features and target variables for the model.



```python
# Create the target variable (tomorrow's close price)
data['Target'] = data['Close'].shift(-1)

# Drop the last row as it will have a NaN target
data.dropna(inplace=True)

# Define the features and target
features = ['Open', 'High', 'Low', 'Close', 'Volume', 'MA_5', 'MA_10', 'MA_50']
target = 'Target'

X = data[features]
y = data[target]

# Display the first few rows of the features and target
display(X.head())
display(y.head())
```

## Split data

### Subtask:
Split the data into training and testing sets using a 70/30 ratio.


**Reasoning**:
Split the data into training and testing sets using a 70/30 ratio.



```python
from sklearn.model_selection import train_test_split

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Print the shapes of the resulting sets
print("Shape of X_train:", X_train.shape)
print("Shape of X_test:", X_test.shape)
print("Shape of y_train:", y_train.shape)
print("Shape of y_test:", y_test.shape)
```

## Build and train regression model

### Subtask:
Choose and train a regression model (e.g., Linear Regression) on the training data.


**Reasoning**:
Train a Linear Regression model using the training data.



```python
from sklearn.linear_model import LinearRegression

# Instantiate the Linear Regression model
model = LinearRegression()

# Fit the model to the training data
model.fit(X_train, y_train)
```

## Evaluate model

### Subtask:
Predict on the test data and report performance metrics such as Mean Squared Error (MSE) and R-squared (R²).


**Reasoning**:
Make predictions on the test set and calculate the MSE and R-squared metrics to evaluate the model's performance.



```python
from sklearn.metrics import mean_squared_error, r2_score

# Make predictions on the test data
y_pred = model.predict(X_test)

# Calculate Mean Squared Error (MSE)
mse = mean_squared_error(y_test, y_pred)
print(f"Mean Squared Error (MSE): {mse}")

# Calculate R-squared (R²) score
r2 = r2_score(y_test, y_pred)
print(f"R-squared (R²) Score: {r2}")
```
